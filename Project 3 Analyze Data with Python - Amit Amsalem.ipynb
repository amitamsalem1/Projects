```python
f_bpm = gym["Resting_BPM"][gym["Gender"]=="Female"]
f_freqency = gym["Session_Duration (hours)"][gym["Gender"]=="Female"]
m_bpm = gym["Resting_BPM"][gym["Gender"]=="Male"]
m_freqency = gym["Session_Duration (hours)"][gym["Gender"]=="Male"]
```

```python
male = gym[gym["Gender"] == "Male"]
female = gym[gym["Gender"] == "Female"]
```

:מסקנות מבוססות נתונים

:ראשית, גברים ונשים הראו מגמה מאוד דומה, אך קיימים הבדלים קטנים לפי הנתונים
אצל הגברים נצפתה ירידה חדה יותר בקצב הלב בזמן מנוחה ככל שכמות האימונים עלתה. אצל הנשים הירידה של קצב דפיקות הלב הייתה מתונה יותר ולכן זה עשוי להעיד על גורמים נוספים כמו הרכב גוף או סוג האימון וכו.
מסקנה זו לא בהכרח נכונה כי כמות הגברים שהשתתפה בניסוי גדולה בכמה אחוזים לעומת כמות הנשים שהשתתפו ולכן אי אפשר לאשש באופן מוחלט מסקנה זו.

שנית, סביב 3–4 אימונים בשבוע, קצב הלב בזמן מנוחה הגיע לממוצע בריא מאוד של 65-70 דפיקות לב בדקה בזמן מנוחה
